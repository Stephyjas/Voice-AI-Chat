```python
from IPython.display import HTML, display

display(HTML("""
<!DOCTYPE html>
<html>

<head>

<style>

* {
    box-sizing: border-box;
    font-family: Arial, sans-serif;
}

body {
    margin: 0;
    background:
        radial-gradient(circle at top left, #30115e, transparent 35%),
        radial-gradient(circle at bottom right, #003d4d, transparent 35%),
        #070b14;

    color: white;
}

.voice-app {
    width: 95%;
    max-width: 850px;
    height: 650px;

    margin: 20px auto;

    background: rgba(15, 20, 35, 0.96);

    border: 1px solid #303952;

    border-radius: 28px;

    box-shadow:
        0 0 60px rgba(120, 70, 255, 0.18);

    overflow: hidden;

    display: flex;
    flex-direction: column;
}


/* HEADER */

.header {
    padding: 22px 28px;

    border-bottom: 1px solid #252d42;

    display: flex;
    align-items: center;
    gap: 15px;
}

.logo {
    width: 48px;
    height: 48px;

    border-radius: 16px;

    background:
        linear-gradient(
            135deg,
            #7c4dff,
            #00c9a7
        );

    display: flex;
    align-items: center;
    justify-content: center;

    font-size: 25px;

    box-shadow:
        0 0 25px rgba(124, 77, 255, .4);
}

.title {
    font-size: 21px;
    font-weight: bold;
}

.subtitle {
    color: #8993aa;
    font-size: 12px;
    margin-top: 4px;
}


/* CHAT */

.chat {
    flex: 1;

    padding: 25px;

    overflow-y: auto;
}

.message {
    max-width: 75%;

    padding: 14px 18px;

    margin-bottom: 18px;

    border-radius: 18px;

    line-height: 1.5;

    font-size: 15px;
}

.bot {
    background: #151c2e;

    border: 1px solid #29334d;

    margin-right: auto;

    border-bottom-left-radius: 5px;
}

.user {
    background:
        linear-gradient(
            135deg,
            #7052e8,
            #934fe8
        );

    margin-left: auto;

    border-bottom-right-radius: 5px;
}


/* INPUT */

.input-area {
    padding: 18px;

    border-top: 1px solid #252d42;

    display: flex;

    gap: 10px;
}

.input {
    flex: 1;

    background: #101625;

    border: 1px solid #303a55;

    border-radius: 15px;

    padding: 15px;

    color: white;

    outline: none;

    font-size: 14px;
}

.input:focus {
    border-color: #7c5cff;
}


/* BUTTON */

.send {
    width: 55px;

    border: none;

    border-radius: 15px;

    background:
        linear-gradient(
            135deg,
            #7657ff,
            #a04cff
        );

    color: white;

    font-size: 20px;

    cursor: pointer;
}

.send:hover {
    transform: scale(1.05);
}


/* VOICE */

.voice {
    margin-top: 8px;

    padding: 6px 12px;

    border: 1px solid #34405d;

    border-radius: 10px;

    background: #101827;

    color: #9ea9c1;

    cursor: pointer;

    font-size: 12px;
}

.voice:hover {
    background: #1c2740;

    color: white;
}


/* FOOTER */

.footer {
    text-align: center;

    padding: 8px;

    color: #5e6981;

    font-size: 11px;
}

</style>

</head>


<body>


<div class="voice-app">


    <!-- HEADER -->

    <div class="header">

        <div class="logo">
            ✦
        </div>

        <div>

            <div class="title">
                VoxAI
            </div>

            <div class="subtitle">
                TEXT → VOICE AI CHAT
            </div>

        </div>

    </div>


    <!-- CHAT -->

    <div class="chat" id="chat">

        <div class="message bot">

            👋 Hello! I'm <b>VoxAI</b>.

            <br><br>

            Type a message below and I'll convert
            the response into voice.

            <br>

            <button
                class="voice"
                onclick="speak(
                    'Hello! I am VoxAI. Welcome to the text to voice chatbot.'
                )">

                🔊 Hear me

            </button>

        </div>

    </div>


    <!-- INPUT -->

    <div class="input-area">

        <input
            id="input"
            class="input"
            placeholder="Type your message..."
            onkeydown="handleKey(event)"
        >

        <button
            class="send"
            onclick="sendMessage()">

            ➤

        </button>

    </div>


    <div class="footer">

        ✦ VoxAI • Text to Voice

    </div>


</div>


<script>


function sendMessage() {

    let input =
        document.getElementById("input");

    let text =
        input.value.trim();


    if (!text) {

        return;

    }


    let chat =
        document.getElementById("chat");


    // USER MESSAGE

    let user =
        document.createElement("div");

    user.className =
        "message user";

    user.innerText =
        text;

    chat.appendChild(user);


    // BOT RESPONSE

    let bot =
        document.createElement("div");

    bot.className =
        "message bot";


    let response =
        "I received your message: " +
        text +
        ". This text can be converted into voice using the speaker button below.";


    bot.innerHTML =
        response +
        "<br><button class='voice' onclick='speak(`" +
        response.replace(/`/g, "'") +
        "`)'>🔊 Speak</button>";


    chat.appendChild(bot);


    input.value = "";


    chat.scrollTop =
        chat.scrollHeight;

}


function handleKey(event) {

    if (event.key === "Enter") {

        sendMessage();

    }

}


function speak(text) {

    if (!("speechSynthesis" in window)) {

        alert(
            "Text-to-speech is not supported in this browser."
        );

        return;

    }


    window.speechSynthesis.cancel();


    let speech =
        new SpeechSynthesisUtterance(text);


    speech.lang =
        "en-US";

    speech.rate =
        0.95;

    speech.pitch =
        1;


    window.speechSynthesis.speak(
        speech
    );

}

</script>


</body>

</html>
"""))
```
